```python
def find_highest(lst):
    if len(lst) == 1:
        return lst[0]
				
    highest_of_rest = find_highest(lst[1:])
    return lst[0] if lst[0] > highest_of_rest else highest_of_rest
```

```python
Test.assert_equals(find_highest([8]), 8)
Test.assert_equals(find_highest([-1, 3, 5, 6, 99, 12, 2]), 99)
Test.assert_equals(find_highest([0, 12, 4, 87]), 87)
```

```python
def number_length(num):
	count = 0
	
	if(num == 0):
		return 1
		
	while(num != 0):
		num	//= 10
		count+=1
	return count	
```

```python
Test.assert_equals(number_length(10), 2)
Test.assert_equals(number_length(5000), 4)
Test.assert_equals(number_length(0), 1)
Test.assert_equals(number_length(4039182), 7)
Test.assert_equals(number_length(9999999999999999), 16)
Test.assert_equals(number_length(1), 1)
Test.assert_equals(number_length(777777777777777777777777777777), 30)
```

```python
def atbash(txt):
    result = []
    for ch in txt:
        if 65 <= ord(ch) <= 90:
            result.append(chr(90 - (ord(ch) - 65)))
        elif 97 <= ord(ch) <= 122:
            result.append(chr(122 - (ord(ch) - 97)))
        else:
            result.append(ch)
    return "".join(result)
```

```python
Test.assert_equals(atbash("abcdefghijklmnopqrstuvwxyz"), "zyxwvutsrqponmlkjihgfedcba")
Test.assert_equals(atbash("ABCDEFGHIJKLMNOPQRSTUVWXYZ"), "ZYXWVUTSRQPONMLKJIHGFEDCBA")
Test.assert_equals(atbash("The word 'atbash' derives from the the first and last 2 letters of the Hebrew alphabet."), "Gsv dliw 'zgyzhs' wvirevh uiln gsv gsv urihg zmw ozhg 2 ovggvih lu gsv Svyivd zokszyvg.")
Test.assert_equals(atbash("Vmxibkgrlm zmw wvxibkgrlm ziv rwvmgrxzo uli gsv Zgyzhs xrksvi."),"Encryption and decryption are identical for the Atbash cipher.")
```

```python
import math
from functools import reduce

def gcd(a, b):
    while b != 0:
        a, b = b, a % b
    return a

def lcm_p(a, b):
    return abs(a * b) // gcd(a, b)

def lcm(nums):
    return reduce(lcm_p, nums)
	
```

```python
Test.assert_equals(lcm([1]), 1)
Test.assert_equals(lcm([5, 5, 5]), 5)
Test.assert_equals(lcm([67, 34, 12, 3, 2]), 13668)
Test.assert_equals(lcm([79, 18, 7, 3, 1]), 9954)
Test.assert_equals(lcm([10, 20, 30, 40, 50]), 600)
Test.assert_equals(lcm([2, 3, 5, 7, 11, 13, 17]), 510510)
Test.assert_equals(lcm([91, 92, 93, 94, 95]), 3476431140)
```
