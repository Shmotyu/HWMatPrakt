```python
#Задача 6. Plot the change in the hare population and in the lynx population for each year. Find the correlation coefficient between the both time series.
```

```python
hd  = np.diff(hares)
ld = np.diff(lynxes)

mean_h = np.mean(hd)
mean_l = np.mean(ld)

cov = np.sum((hd - mean_h) * (ld - mean_l)) / (len(hd) - 1)

std_h = np.sqrt(np.sum((hd - mean_h)**2) / (len(hd) - 1))
std_l = np.sqrt(np.sum((ld - mean_l)**2) / (len(ld) - 1))

cor = cov / (std_h * std_l)

print("Correlation =", cor)

plt.figure(figsize=(12, 6))

plt.plot(year[1:], hd, label='Hare change')
plt.plot(year[1:], ld, label='Lynx change')

plt.xlabel('Year')
plt.ylabel('Population change')
plt.title('Yearly changes in hare and lynx populations')
plt.legend()
plt.grid(True)
plt.xticks(year[1:])
plt.show()
```

```python
#Задание 9:

# 1. напишите функцию, реализующую алгоритм k-means.
# Входом в функцию является

# массив NxM, где N -- число объектов, M -- число признаков объекта
# k -- число кластеров
# max_iter -- максимальное число итераций
# threshold -- критерий остановки (если на очередной итерации метода суммарное квадратичное отклонение изменяется меньше, чем на threshold, итерации прекращаются)
# Выход:

# массив длины N, содержащий индексы кластеров от 0 до k-1
# суммарное квадратичное отклонение для полученного разбиения на кластеры
# Постарайтесь обойтись минимальным количеством циклов в вашем коде.
```

```python
def k_means(arr, k, max_iter, threshold):
    #Выберем в качестве центров первые k объекта
    centers = arr[:k].copy()
    index = np.empty(arr.shape[0], dtype=int)
    sse = 0
    
    for i in range(max_iter):
        #Для каждого j находим k: ||arr[j] - centers[k]|| -> min
        for j in range(arr.shape[0]):
            index[j] = np.argmin(np.linalg.norm(arr[j] - centers, axis=1))
        
        #найдем новые центры
        for j in range(k):
            mask = (index == j)
            centers[j] = arr[mask].mean(axis=0)
        
        #Вычислим суммарное квадратичное отклонение
        sse1 = 0
        for j in range(arr.shape[0]):
            sse1 += np.linalg.norm(arr[j] - centers[index[j]])**2
        
        if i != 0 and abs(sse - sse1) < threshold:
            break
        sse = sse1
    
    return index, sse, centers

```

```python
# Задание 10:
# 2. Выделите 10 кластеров в датасете MNIST. Дайте интерпретацию каждому полученному кластеру (к какой цифре его можно отнести).
# Нарисуйте изображения, которые отвечают центрам кластеров. Посчитайте, сколько из точек кластера действительно являются той цифрой,
# к которой вы его отнесли (для этого сравните с массивом правильных ответов labels) и нарисуйте матрицу ошибок.
```

```python
from sklearn.datasets import fetch_openml

#Загрузка может занять около минуты
images, labels = fetch_openml('mnist_784', return_X_y=True, as_frame=False, parser='auto')
```

```python
from sklearn.metrics import confusion_matrix

labels = labels.astype(int)  
images = images / 255.0

k = 10
max_iter = 10  
threshold = 0

index, sse, centers = k_means(images, k, max_iter, threshold)
print("Итоговый SSE:", sse)


fig, axes = plt.subplots(1, 10, figsize=(15, 3))
for i, ax in enumerate(axes):
    ax.imshow(centers[i].reshape(28, 28), cmap='Greys_r')
    ax.set_title(f'Cluster {i}')
    ax.axis('off')
plt.suptitle("Центры кластеров k-means")
plt.show()

cluster_to_digit = {}
print("\nИнтерпретация кластеров:")
for c in range(k):
    cluster_mask = (index == c)
    cluster_labels = labels[cluster_mask]
    if cluster_labels.size == 0:
        print(f"Кластер {c}: пустой")
        continue

    values, counts = np.unique(cluster_labels, return_counts=True)
    majority_digit = values[np.argmax(counts)]
    cluster_to_digit[c] = majority_digit

    print(f"Кластер {c}: чаще всего цифра {majority_digit} "
          f"({np.max(counts)} из {cluster_labels.size})")


print("\nДоля верных цифр в каждом кластере:")
for c, digit in cluster_to_digit.items():
    mask = (index == c)
    cluster_labels = labels[mask]
    correct = np.sum(cluster_labels == digit)
    total = cluster_labels.size
    acc = correct / total if total > 0 else 0
    print(f"Кластер {c} → цифра {digit}: {correct}/{total} "
          f"({acc:.2%} верных)")


predicted_digits = np.zeros_like(labels)
for c, digit in cluster_to_digit.items():
    predicted_digits[index == c] = digit

cm = confusion_matrix(labels, predicted_digits, labels=np.arange(10))

print("\nМатрица ошибок (строки — истинные цифры, столбцы — предсказанные):\n")

for i in range(10):
    row = " ".join([f"{cm[i, j]:5d}" for j in range(10)])
    print(f"{i:2d} | {row}")
```
